# Step 9 - Final Results Summary

## Dataset choice

Selected dataset: GSE305638 - Frataxin deficiency drives cardiac dysfunction and transcriptional dysregulation in Friedreich ataxia iPSC model. This is a Friedreich ataxia / frataxin-related mitochondrial disease model using human iPSC-derived cardiomyocytes. It is not MERRF-specific; it is used as a related mitochondrial cardiomyopathy model because it combines human iPSC-cardiomyocytes, mitochondrial disease biology, cardiac dysfunction, processed RNA-seq counts, and paired disease/control metadata.

## Methods

Metadata curation produced 9 disease and 9 control samples across 3 patient/isogenic-control pairs. Expression QC aligned 18 samples and 59743 gene rows with 0 missing count values. Exploratory paired expression analysis used log2 CPM normalization, replicate averaging within patient/group, paired disease-control tests across three patient pairs, and Benjamini-Hochberg FDR correction.

## Main expression findings

The exploratory comparison tested 15100 expressed genes. 859 genes had nominal p < 0.05, but 0 genes passed FDR < 0.10 and 0 passed FDR < 0.05. This supports reporting ranked exploratory signals rather than claiming genome-wide findings after FDR correction.

## Mitochondrial gene findings

The mitochondrial reference contains 41 genes, with 40 tested in the exploratory expression table. The minimum mitochondrial adjusted p-value was 0.671, so no mitochondrial gene is FDR-significant. Descriptively, mtDNA-encoded OXPHOS genes trend lower in disease; the strongest nominal mitochondrial genes include POLG, ATP5F1A, MT-ND2, MT-CO3, PPARGC1A.

## Limitations

The selected dataset models Friedreich ataxia/frataxin deficiency, not MERRF or MT-TK m.8344A>G. The exploratory paired expression analysis is limited by the effective paired sample size of three patient pairs and the Python implementation is not a full DESeq2 negative-binomial model. Mitochondrial pathway summaries are descriptive selected-gene summaries, not formal enrichment tests.
